## How to Calculate Fidelity between 2 QROMs?
Fidelity is often computed with no noise at all to determine how logically equivalent two circuits are.

Noise-added fidelity is what people often do for hardware studies. The fidelity drop is caused by:

depolarizing noise
amplitude damping
readout errors
etc.

This measures hardware robustness.

The "ideal QROM" is not a particular circuit architecture.

It is the desired unitary transformation:

$∣a⟩∣0⟩→∣a⟩∣d(a)⟩$

where d(a) is the stored data.

### First try with sampling each index takes too long

```python
def evaluate_QROM(bitstrings, num_index_bits, num_swap_bits=0, bits_per_value=2, print_out=False):

    if(len(bitstrings) != 2**num_index_bits or len(bitstrings[0]) != bits_per_value):
        print(f"Error: Mismatch in size of bitstrings and expected number of num_index_bits and bits_per_value. ({len(bitstrings)}, {len(bitstrings[0])}) != ({2**num_index_bits}, {bits_per_value})")
        return -1, -1, -1, -1, -1
    
    control_wires = np.arange(num_index_bits)
    target_wires  = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires    = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))
    
    @partial(qp.compile, basis_set="CNOT")  # Line added for resource estimation purposes only.
    @qp.set_shots(1)
    @qp.qnode(qp.device("default.qubit"))
    def circuit(index):
        qp.BasisState(index, wires=control_wires)
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)
        return qp.sample(wires=target_wires), qp.sample(wires=target_wires)

    fidelities = []
    full_start_time = time.perf_counter()
    
    for index in range(2**num_index_bits):

        start_time = time.perf_counter()

        # Actually execute the QROM
        samples = circuit(index)

        # Ideal output bitstring
        ideal_output = np.array(list(map(int, bitstrings[index])))

        # Check each shot against ideal target
        correct = np.all(samples == ideal_output, axis=1)

        # Fraction of shots giving correct answer
        fidelity = np.mean(correct)

        elapsed = time.perf_counter() - start_time

        if print_out: print(f"Retrieval Fidelity for Index {index} is {fidelity} and took {elapsed:.6f} s")
        fidelities.append(float(fidelity))

    
    r = qp.specs(circuit)(0)["resources"]

    num_qubits = len(control_wires) + len(target_wires) + len(work_wires)
    oneQ = r.gate_sizes.get(1,0)
    twoQ = r.gate_sizes.get(2,0)

    if print_out:
        print("Number of qubits: ", num_qubits)
        print("One-qubit gates: ", oneQ)
        print("Two-qubit gates: ", twoQ)
        print(f"Depth: {r.depth}")

    return {
        "swap_bits": num_swap_bits,
        "qubits": num_qubits,
        "oneQ": oneQ,
        "twoQ": twoQ,
        "depth": r.depth,
    }
```

```python
def evaluate_QROM(bitstrings, num_index_bits, num_swap_bits=0, bits_per_value=2, print_out=False):

    if(len(bitstrings) != 2**num_index_bits or len(bitstrings[0]) != bits_per_value):
        print(f"Error: Mismatch in size of bitstrings and expected number of num_index_bits and bits_per_value. ({len(bitstrings)}, {len(bitstrings[0])}) != ({2**num_index_bits}, {bits_per_value})")
        return -1, -1, -1, -1, -1
    
    control_wires = np.arange(num_index_bits)
    target_wires  = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires    = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))
    
    @partial(qp.compile, basis_set="CNOT")
    @qp.qnode(qp.device("default.qubit"))
    def circuit(index):
        qp.BasisState(index, wires=control_wires)
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)
    
        return qp.density_matrix(wires=list(control_wires) + list(target_wires))
    
    
    fidelities = []
    full_start_time = time.perf_counter()
    
    for index in range(2**num_index_bits):
    
        start_time = time.perf_counter()
    
        # Actual control + target density matrix
        actual_state = circuit(index)
    
        # Ideal state = |index>|bitstrings[index]>
        ideal_bits = (list(map(int, f"{index:0{num_index_bits}b}")) + list(map(int, bitstrings[index])))
        ideal_index = int("".join(map(str, ideal_bits)), 2)
    
        # Ideal statevector
        ideal_statevector = np.zeros(2**(num_index_bits + bits_per_value),dtype=complex)
        ideal_statevector[ideal_index] = 1.0
    
        # Convert ideal statevector to density matrix
        ideal_state = np.outer(ideal_statevector,np.conj(ideal_statevector))

        # True density-matrix fidelity
        fidelity = qp.math.fidelity(actual_state,ideal_state)
        fidelity = np.clip(np.real(fidelity), 0.0, 1.0)
    
        elapsed = time.perf_counter() - start_time

        
        if print_out: print(f"Retrieval Fidelity for Index {index} is {fidelity} and took {elapsed:.6f} s")
        fidelities.append(float(fidelity))

        del actual_state
        del ideal_state
        del ideal_statevector
        del ideal_bits
    
    full_elapsed = time.perf_counter() - full_start_time
    retrieval_fidelity = np.mean(fidelities)
    
    if print_out:
        print(
            f"Final Retrieval Fidelity is {retrieval_fidelity} "
            f"and took {full_elapsed:.6f} s"
        )

    r = qp.specs(circuit)(0)["resources"]

    num_qubits = len(control_wires) + len(target_wires) + len(work_wires)
    oneQ = r.gate_sizes.get(1,0)
    twoQ = r.gate_sizes.get(2,0)

    if print_out:
        print("Number of qubits: ", num_qubits)
        print("One-qubit gates: ", oneQ)
        print("Two-qubit gates: ", twoQ)
        print(f"Depth: {r.depth}")

    return {
        "swap_bits": num_swap_bits,
        "qubits": num_qubits,
        "oneQ": oneQ,
        "twoQ": twoQ,
        "depth": r.depth,
    }
```

Resulting errors above 1 are caused by floating point precision of near identical values. Yes. **Memory Requirement**: Statevector Fidelity has a memory requirement of 16(2^N). Density-matrix fidelity requires 16(4^N) for a density matrix of 2^n x 2^n. SWAP test reqruires 2n+1 states, so 16(2^(2N+1)) byte memory requirement. 16 is there for a complex128 number. 

**Time Complexity**: O(2^N) for statevector fidelity, O(2^{3n}) for density-matrix fidelity, SWAPtest has cheap measurement

# Extra Work

### Manually Running through the Gates

```python
def apply_gate(bits, gate):

    wires = list(gate.wires)

    if gate.name == "PauliX":
        bits[wires[0]] ^= 1

    elif gate.name == "CNOT":
        control, target = wires
        bits[target] ^= bits[control]

    elif gate.name == "SWAP":
        a, b = wires
        bits[a], bits[b] = bits[b], bits[a]

    elif gate.name == "Toffoli":
        control1, control2, target = wires

        if bits[control1] and bits[control2]:
            bits[target] ^= 1

    elif gate.name == "MultiControlledX":
        controls = wires[:-1]
        target = wires[-1]

        if all(bits[c] for c in controls):
            bits[target] ^= 1

    else:
        raise ValueError(f"Unsupported gate: {gate.name}")

    return bits
```

```python
def evaluate_QROM(bitstrings, num_index_bits, num_swap_bits=0, bits_per_value=2, print_out=False):

    if(len(bitstrings) != 2**num_index_bits or len(bitstrings[0]) != bits_per_value):
        print(f"Error: Mismatch in size of bitstrings and expected number of num_index_bits and bits_per_value. ({len(bitstrings)}, {len(bitstrings[0])}) != ({2**num_index_bits}, {bits_per_value})")
        return -1, -1, -1, -1, -1

    full_start_time = time.perf_counter()
    
    control_wires = np.arange(num_index_bits)
    target_wires = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))
    
    def circuit():
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)

    # Decompose QROM while keeping classical reversible gates
    gate_set = {"PauliX", "CNOT", "SWAP", "MultiControlledX", "Toffoli"}
    decomposed_circuit = qp.transforms.decompose(circuit, gate_set=gate_set)
    tape = qp.tape.make_qscript(decomposed_circuit)()
    gate_list = tape.operations
    
    fidelities = []

    num_qubits = len(control_wires) + len(target_wires) + len(work_wires)
    
    for index in range(2**num_index_bits):
        start_time = time.perf_counter()

        # Store one bit per qubit instead of a statevector
        actual_state = np.zeros(num_qubits, dtype=np.uint8)

        # Initial control state
        ideal_bits = list(map(int, f"{index:0{num_index_bits}b}"))
        actual_state[control_wires] = ideal_bits

        # Apply QROM gates manually
        for gate in gate_list:
            actual_state = apply_gate(actual_state, gate)

        # Actual target output
        actual_data = actual_state[target_wires]

        # Ideal target output
        ideal_state = np.array(list(map(int, bitstrings[index])))

        # Logical retrieval fidelity
        fidelity = float(np.array_equal(actual_data, ideal_state))
    
        elapsed = time.perf_counter() - start_time

        if print_out: print(f"Retrieval Fidelity for Index {index} is {fidelity} and took {elapsed:.6f} s", flush=True)
        fidelities.append(float(fidelity))

        del actual_state
        del actual_data
        del ideal_state
        del ideal_bits
    
    retrieval_fidelity = np.mean(fidelities)
    full_elapsed = time.perf_counter() - full_start_time
    if print_out: print(f"Final Retrieval Fidelity is {retrieval_fidelity} and took {full_elapsed:.6f} s")

    oneQ = sum(1 for gate in gate_list if len(gate.wires) == 1)
    twoQ = sum(1 for gate in gate_list if len(gate.wires) == 2)

    if print_out:
        print("Number of qubits: ", num_qubits)
        print("One-qubit gates: ", oneQ)
        print("Two-qubit gates: ", twoQ)
    
    return {
        "swap_bits": num_swap_bits,
        "qubits": num_qubits,
        "oneQ": oneQ,
        "twoQ": twoQ,
        "fidelity": retrieval_fidelity,
    }
```

### Manually Running through the Gates (with Decimals)

```python
def evaluate_QROM(bitstrings, num_index_bits, num_swap_bits=0, bits_per_value=2, print_out=False):

    if(len(bitstrings) != 2**num_index_bits or len(bitstrings[0]) != bits_per_value):
        print(f"Error: Mismatch in size of bitstrings and expected number of num_index_bits and bits_per_value. ({len(bitstrings)}, {len(bitstrings[0])}) != ({2**num_index_bits}, {bits_per_value})")
        return -1, -1, -1, -1, -1
    
    control_wires = np.arange(num_index_bits)
    target_wires = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))

    num_qubits = len(control_wires) + len(target_wires) + len(work_wires)

    dev = qp.device("default.tensor", wires=num_qubits, method="mps", max_bond_dim=64, cutoff=1e-12)

    @partial(qp.compile, basis_set="CNOT")
    @qp.qnode(dev)
    def circuit(index):
        qp.BasisState(index, wires=control_wires)
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)
        return [qp.expval(qp.Z(wire)) for wire in list(control_wires) + list(target_wires)]
    
    fidelities = []
    full_start_time = time.perf_counter()
    
    for index in range(2**num_index_bits):
        start_time = time.perf_counter()

        actual_state = np.array(circuit(index))

        ideal_bits = list(map(int, f"{index:0{num_index_bits}b}")) + list(map(int, bitstrings[index]))

        fidelities_per_bit = []
        for z, bit in zip(actual_state, ideal_bits):
            probability = (1 + z)/2 if bit == 0 else (1 - z)/2
            fidelities_per_bit.append(probability)

        fidelity = np.prod(fidelities_per_bit)
        fidelity = np.clip(np.real(fidelity), 0.0, 1.0)

        elapsed = time.perf_counter() - start_time

        if print_out: print(f"Retrieval Fidelity for Index {index} is {fidelity:.15f} and took {elapsed:.6f} s", flush=True)
        fidelities.append(float(fidelity))

    full_elapsed = time.perf_counter() - full_start_time
    retrieval_fidelity = np.mean(fidelities)

    if print_out: print(f"Final Retrieval Fidelity is {retrieval_fidelity:.15f} and took {full_elapsed:.6f} s")

    r = qp.specs(circuit)(0)["resources"]

    oneQ = r.gate_sizes.get(1,0)
    twoQ = r.gate_sizes.get(2,0)

    if print_out:
        print("Number of qubits: ", num_qubits)
        print("One-qubit gates: ", oneQ)
        print("Two-qubit gates: ", twoQ)
        print(f"Depth: {r.depth}")

    return {
        "swap_bits": num_swap_bits,
        "qubits": num_qubits,
        "oneQ": oneQ,
        "twoQ": twoQ,
        "depth": r.depth,
        "fidelity": retrieval_fidelity,
    }
```

### MPS without truncation

```python
def evaluate_QROM(bitstrings, num_index_bits, num_swap_bits=0, bits_per_value=2, print_out=False):

    if(len(bitstrings) != 2**num_index_bits or len(bitstrings[0]) != bits_per_value):
        print(f"Error: Mismatch in size of bitstrings and expected number of num_index_bits and bits_per_value. ({len(bitstrings)}, {len(bitstrings[0])}) != ({2**num_index_bits}, {bits_per_value})")
        return -1, -1, -1, -1, -1
    
    control_wires = np.arange(num_index_bits)
    target_wires = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))

    num_qubits = len(control_wires) + len(target_wires) + len(work_wires)

    dev = qp.device("default.tensor", wires=num_qubits, method="mps")

    @partial(qp.compile, basis_set="CNOT")
    @qp.qnode(dev)
    def circuit(index):

        qp.BasisState(index, wires=control_wires)
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)

        ideal_bits = list(map(int, f"{index:0{num_index_bits}b}")) + list(map(int, bitstrings[index]))
        ideal_wires = list(control_wires) + list(target_wires)

        return qp.expval(qp.Projector(ideal_bits, wires=ideal_wires))
    
    fidelities = []
    full_start_time = time.perf_counter()
    
    for index in range(2**num_index_bits):
        start_time = time.perf_counter()

        fidelity = circuit(index)
        fidelity = np.clip(np.real(fidelity), 0.0, 1.0)

        elapsed = time.perf_counter() - start_time

        if print_out: print(f"Retrieval Fidelity for Index {index} is {fidelity:.15f} and took {elapsed:.6f} s", flush=True)
        fidelities.append(float(fidelity))
    
    full_elapsed = time.perf_counter() - full_start_time
    retrieval_fidelity = np.mean(fidelities)

    if print_out: print(f"Final Retrieval Fidelity is {retrieval_fidelity:.15f} and took {full_elapsed:.6f} s")

    r = qp.specs(circuit)(0)["resources"]

    oneQ = r.gate_sizes.get(1,0)
    twoQ = r.gate_sizes.get(2,0)

    if print_out:
        print("Number of qubits: ", num_qubits)
        print("One-qubit gates: ", oneQ)
        print("Two-qubit gates: ", twoQ)
        print(f"Depth: {r.depth}")

    return {
        "swap_bits": num_swap_bits,
        "qubits": num_qubits,
        "oneQ": oneQ,
        "twoQ": twoQ,
        "depth": r.depth,
        "fidelity": retrieval_fidelity,
    }
```

A Matrix Product State (MPS) is a compressed way of representing a quantum state without explicitly storing all 2N2^N2N amplitudes of the full statevector. Instead, it represents the state as a chain of smaller tensors, with the bond dimension controlling how much entanglement and information can be retained. In our QROM simulation, limiting the bond dimension and applying a cutoff allows the MPS to approximate the full quantum state using substantially less memory, and we can then compute the retrieval fidelity directly from that approximate state.

The full statevector stores all 2N2^N2N amplitudes explicitly.
The MPS represents that same many-qubit state as a chain of smaller tensors.

### MPS with Truncation but more extreme

```python
def evaluate_QROM(bitstrings, num_index_bits, num_swap_bits=0, bits_per_value=2, print_out=False):

    if(len(bitstrings) != 2**num_index_bits or len(bitstrings[0]) != bits_per_value):
        print(f"Error: Mismatch in size of bitstrings and expected number of num_index_bits and bits_per_value. ({len(bitstrings)}, {len(bitstrings[0])}) != ({2**num_index_bits}, {bits_per_value})")
        return -1, -1, -1, -1, -1
    
    control_wires = np.arange(num_index_bits)
    target_wires = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))

    num_qubits = len(control_wires) + len(target_wires) + len(work_wires)

    dev = qp.device("default.tensor", wires=num_qubits, method="mps", contraction_optimizer="greedy",
                    max_bond_dim=64, cutoff=1e-12)

    @partial(qp.compile, basis_set="CNOT")
    @qp.qnode(dev)
    def circuit(index):
        qp.BasisState(index, wires=control_wires)
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)
        ideal_bits = list(map(int, f"{index:0{num_index_bits}b}")) + list(map(int, bitstrings[index]))
        ideal_wires = list(control_wires) + list(target_wires)
        return qp.expval(qp.Projector(ideal_bits, wires=ideal_wires))
    
    fidelities = []
    full_start_time = time.perf_counter()
    
    for index in np.linspace(0, 2**num_index_bits - 1, num=10, dtype=int):
        start_time = time.perf_counter()

        fidelity = circuit(index)
        fidelity = np.clip(np.real(fidelity), 0.0, 1.0)

        elapsed = time.perf_counter() - start_time

        if print_out: print(f"Retrieval Fidelity for Index {index} is {fidelity:.15f} and took {elapsed:.6f} s", flush=True)
        fidelities.append(float(fidelity))
    
    full_elapsed = time.perf_counter() - full_start_time
    retrieval_fidelity = np.mean(fidelities)

    if print_out: print(f"Final Retrieval Fidelity is {retrieval_fidelity:.15f} and took {full_elapsed:.6f} s")

    r = qp.specs(circuit)(0)["resources"]

    oneQ = r.gate_sizes.get(1,0)
    twoQ = r.gate_sizes.get(2,0)

    if print_out:
        print("Number of qubits: ", num_qubits)
        print("One-qubit gates: ", oneQ)
        print("Two-qubit gates: ", twoQ)
        print(f"Depth: {r.depth}")

    return {
        "swap_bits": num_swap_bits,
        "qubits": num_qubits,
        "oneQ": oneQ,
        "twoQ": twoQ,
        "depth": r.depth,
        "fidelity": retrieval_fidelity,
    }
```

### MPS + apply gates manually

```python
import pennylane as qp
import quimb.tensor as qtn
import numpy as np
import time


def decompose_QROM(bitstrings, control_wires, target_wires, work_wires):
    ops = [qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)]

    while any(len(op.wires) > 2 for op in ops):
        new_ops = []

        for op in ops:
            if len(op.wires) <= 2:
                new_ops.append(op)
            else:
                new_ops.extend(op.decomposition())

        ops = new_ops

    return ops


def evaluate_QROM(bitstrings, num_index_bits, num_swap_bits=0, bits_per_value=2, max_bond=32, cutoff=1e-8, print_out=False):

    control_wires = np.arange(num_index_bits)
    target_wires = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))

    num_qubits = len(control_wires) + len(target_wires) + len(work_wires)
    ops = decompose_QROM(bitstrings, control_wires, target_wires, work_wires)

    fidelities = []

    for index in np.linspace(0, 2**num_index_bits - 1, 10, dtype=int):
        start = time.perf_counter()

        mps = qtn.CircuitMPS(N=num_qubits, max_bond=max_bond, cutoff=cutoff)

        index_bits = format(index, f"0{num_index_bits}b")

        for wire, bit in zip(control_wires, index_bits):
            if bit == "1":
                mps.apply_gate("X", int(wire))

        for op in ops:
            mps.apply_gate(np.asarray(qp.matrix(op)), *map(int, op.wires))

        ideal = ["0"] * num_qubits

        for wire, bit in zip(control_wires, index_bits):
            ideal[int(wire)] = bit

        for wire, bit in zip(target_wires, bitstrings[index]):
            ideal[int(wire)] = bit

        fidelity = abs(mps.amplitude("".join(ideal)))**2
        fidelity = float(np.clip(np.real(fidelity), 0, 1))
        fidelities.append(fidelity)

        if print_out:
            print(f"{index}: F={fidelity:.8f}, t={time.perf_counter()-start:.2f}s")

    return {"swap_bits": num_swap_bits, "qubits": num_qubits, "fidelity": np.mean(fidelities)}
```

For the fidelity calculation, we only are targeting the control and target wires, not the work wires because the intermediate results don't matter as long as the input and outputs are good.

```python
def QROM_fidelity(bitstrings, num_index_bits, num_swap_bits=0, bits_per_value=2, print_out=False):

    if(len(bitstrings) != 2**num_index_bits or len(bitstrings[0]) != bits_per_value):
        print(f"Error: Mismatch in size of bitstrings and expected number of num_index_bits and bits_per_value. ({len(bitstrings)}, {len(bitstrings[0])}) != ({2**num_index_bits}, {bits_per_value})")
        return -1
    
    control_wires = np.arange(num_index_bits)
    target_wires  = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires    = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))
    
    @partial(qp.compile, basis_set="CNOT")  # Line added for resource estimation purposes only.
    @qp.set_shots(1)
    @qp.qnode(qp.device("default.qubit"))
    def circuit(index):
        qp.BasisState(index, wires=control_wires)
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)
        return qp.sample(wires=target_wires)

    fidelities = []
    full_start_time = time.perf_counter()
    
    for index in range(2**num_index_bits):

        start_time = time.perf_counter()
        # Actual control + target state
        actual_state = circuit(index)
    
        # Ideal state = |index>|bitstrings[index]>
        ideal_bits = (list(map(int, f"{index:0{num_index_bits}b}")) + list(map(int, bitstrings[index])))
        ideal_index = int("".join(map(str, ideal_bits)), 2)
    
        # Ideal statevector
        ideal_statevector = np.zeros(2**(num_index_bits + bits_per_value), dtype=complex)
        ideal_statevector[ideal_index] = 1.0
    
        # Convert to density matrix
        ideal_state = np.outer(ideal_statevector, np.conj(ideal_statevector))
    
        # Fidelity
        fidelity = qp.math.fidelity(actual_state, ideal_state)

        elapsed = time.perf_counter() - start_time
        print(f"Retrieval Fidelity for Index {index} is {fidelity} and took {elapsed:.6f} s")
        fidelities.append(float(fidelity))

    full_elapsed = time.perf_counter() - full_start_time
    retrieval_fidelity = np.mean(fidelities)
    print(f"Final Retrieval Fidelity is {retrieval_fidelity} and took {full_elapsed:.6f} s")
```

```python
def QROM_fidelity_test(bitstrings, num_index_bits, num_swap_bits=0,
                       bits_per_value=2, shots=1000, print_out=False):

    if(len(bitstrings) != 2**num_index_bits or len(bitstrings[0]) != bits_per_value):
        print(f"Error: Mismatch in size of bitstrings and expected number of num_index_bits and bits_per_value. ({len(bitstrings)}, {len(bitstrings[0])}) != ({2**num_index_bits}, {bits_per_value})")
        return -1

    control_wires = np.arange(num_index_bits)
    target_wires  = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires    = np.arange(
        target_wires[-1] + 1,
        target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1)
    )

    num_qubits = (
        len(control_wires)
        + len(target_wires)
        + len(work_wires)
    )

    # Original QROM circuit
    @partial(qp.compile, basis_set="CNOT")
    @qp.set_shots(shots)
    @qp.qnode(qp.device("default.qubit"))
    def circuit(index):

        qp.BasisState(index, wires=control_wires)

        qp.QROM(
            bitstrings,
            control_wires,
            target_wires,
            work_wires,
            clean=False
        )

        return qp.sample(wires=target_wires)


    fidelities = []

    full_start_time = time.perf_counter()


    for index in range(2**num_index_bits):

        start_time = time.perf_counter()

        # Actually execute the QROM
        samples = circuit(index)

        # Ideal output bitstring
        ideal_output = np.array(
            list(map(int, bitstrings[index]))
        )

        # Check each shot against ideal target
        correct = np.all(
            samples == ideal_output,
            axis=1
        )

        # Fraction of shots giving correct answer
        fidelity = np.mean(correct)

        elapsed = time.perf_counter() - start_time

        if print_out:
            print(
                f"Retrieval Fidelity for Index {index} "
                f"is {fidelity} and took {elapsed:.6f} s"
            )

        fidelities.append(float(fidelity))


    full_elapsed = time.perf_counter() - full_start_time

    retrieval_fidelity = np.mean(fidelities)


    if print_out:
        print(
            f"Final Retrieval Fidelity is {retrieval_fidelity} "
            f"and took {full_elapsed:.6f} s"
        )

    return retrieval_fidelity
```

```python
# 8x8 Image of uint8 = 2^3 x 2^3 x 8 bits
num_index_bits = 6
bits_per_value = 2
num_swap_bits=5

bitstrings = generate_bitstrings(2**num_index_bits, bits_per_value, worstcase=True)
QROM_fidelity_test(bitstrings, num_index_bits, num_swap_bits, bits_per_value, print_out=False)
```
