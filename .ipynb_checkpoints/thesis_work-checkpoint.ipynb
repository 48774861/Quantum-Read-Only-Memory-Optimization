```python
QROM_size = 64
bits_per_value = 2
num_index_bits = int(np.ceil(np.log2(QROM_size)))

def generate_bitstrings(array_length, bit_length):
    return [f"{random.randint(0, 2**bit_length - 1):0{bit_length}b}" for _ in range(array_length)]

bitstrings = generate_bitstrings(QROM_size, bits_per_value)

swap_bits_list, qubits, oneq, twoq, depth = [], [], [], [], []

for num_swap_bits in range(num_index_bits + 1):

    start = time.perf_counter()
    
    control_wires = np.arange(num_index_bits)
    target_wires  = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires    = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))
    
    @partial(qp.compile, basis_set="CNOT")  # Line added for resource estimation purposes only.
    @qp.set_shots(1)
    @qp.qnode(qp.device("default.qubit"))
    def circuit(index):
        qp.BasisState(index, wires=control_wires)
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)
        return qp.sample(wires=target_wires), qp.sample(wires=target_wires)

    r = qp.specs(circuit)(0)["resources"]

    swap_bits_list.append(num_swap_bits)
    qubits.append(len(control_wires) + len(target_wires) + len(work_wires))
    oneq.append(r.gate_sizes.get(1, 0))
    twoq.append(r.gate_sizes.get(2, 0))
    depth.append(r.depth)
    
    elapsed = time.perf_counter() - start
    print(f"Done: {num_swap_bits} swap bits in {elapsed:.6f} s")
```

### Number of Gates if we work ourselves from the bottom

```python
def draw_qrom(QROM_size, num_swap_bits=0, bits_per_value=2):

    num_index_bits = int(np.ceil(np.log2(QROM_size)))
    
    def generate_bitstrings(array_length, bit_length):
        return [f"{random.randint(0, 2**bit_length - 1):0{bit_length}b}" for _ in range(array_length)]
    
    bitstrings = generate_bitstrings(QROM_size, bits_per_value)
    
    control_wires = np.arange(num_index_bits)
    target_wires  = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires    = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))
    
    # Line added for drawing purposes only
    @partial(qp.transforms.decompose, max_expansion=2)
    @qp.set_shots(1)
    @qp.qnode(qp.device("default.qubit"))
    def circuit(index):
        qp.BasisState(index, wires=control_wires)
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)
        return qp.sample(wires=target_wires), qp.sample(wires=target_wires)
    
    qp.draw_mpl(circuit, style="pennylane")(0)
    plt.show()
```

```python
def analyze_qrom(QROM_size, num_swap_bits=0, bits_per_value=2, print_out=True):

    num_index_bits = int(np.ceil(np.log2(QROM_size)))
    
    def generate_bitstrings(array_length, bit_length):
        return [f"{random.randint(0, 2**bit_length - 1):0{bit_length}b}" for _ in range(array_length)]
    
    bitstrings = generate_bitstrings(QROM_size, bits_per_value)
    
    control_wires = np.arange(num_index_bits)
    target_wires  = np.arange(num_index_bits, num_index_bits + bits_per_value)
    work_wires    = np.arange(target_wires[-1] + 1, target_wires[-1] + 1 + bits_per_value*(2**num_swap_bits - 1))
    
    @partial(qp.compile, basis_set="CNOT")  # Line added for resource estimation purposes only.
    @qp.set_shots(1)
    @qp.qnode(qp.device("default.qubit"))
    def circuit(index):
        qp.BasisState(index, wires=control_wires)
        qp.QROM(bitstrings, control_wires, target_wires, work_wires, clean=False)
        return qp.sample(wires=target_wires), qp.sample(wires=target_wires)
    
    r = qp.specs(circuit)(0)["resources"]

    num_qubits = len(control_wires) + len(target_wires) + len(work_wires)
    oneQ = r.gate_sizes.get(1,0)
    twoQ = r.gate_sizes.get(2,0)

    if print_out:
        print("Number of qubits: ", num_qubits)
        print("One-qubit gates: ", oneQ)
        print("Two-qubit gates: ", twoQ)
        print(f"Depth: {r.depth}")

    return num_qubits, oneQ, twoQ, r.depth
```

```python
def run_analyze_qrom(QROM_size, num_swap_bits=0, bits_per_value=2, print_out=True, N=30):
    qubits_list = []
    oneQ_list = []
    twoQ_list = []
    depth_list = []
    
    for _ in range(N):
        num_qubits, oneQ, twoQ, depth = analyze_qrom(QROM_size, num_swap_bits, bits_per_value, print_out=False)
        qubits_list.append(num_qubits)
        oneQ_list.append(oneQ)
        twoQ_list.append(twoQ)
        depth_list.append(depth)
    
    for name, data in {
        "Qubits": qubits_list,
        "1Q Gates": oneQ_list,
        "2Q Gates": twoQ_list,
        "Depth": depth_list,
    }.items():
        print(
            f"{name:10s} | "
            f"min={np.min(data):.2f} | "
            f"max={np.max(data):.2f} | "
            f"mean={np.mean(data):.2f}"
        )
```

```python
def estimate_gates(QROM_size, num_swap_bits=0, bits_per_value=2):
    oneQ = bits_per_value
    twoQ = 0
    print(f"Estimated 1Q Gates: {oneQ}")
    print(f"Estimated 2Q Gates: {twoQ}")
```

### Number of Gates for QROM_size = 2

Setting the Data: In total, there is QROM_size * bits_per_value number of bits to write. One CNOT is added if the bit being written is 1; nothing is added if the bit being written is 0. Thus, if we have 2 bits per value and a QROM size of 2, writing all 0's would yield no additional CNOTs added.

Depth here when running multiple trials should obviously be 1Q + 2Q since over all trials it would reach that at least once. Even though the target we want may be at the end, the index bit still needs to go through all the X and CNOT gates to get there. For example, QROM_size = 2 and 2 bits per value has 2 X-gates to alternate the index and 4 CNOT gates between index and target bits. Max depth is 6 because the index bit needs to go through all 6 anyway.

Layer 1: X(index)

Layer 2: CNOT(index,target0)

Layer 3: CNOT(index,target1)

Layer 4: X(index)

Layer 5: CNOT(index,target0)

Layer 6: CNOT(index,target1)

Verified that the CNOT gate is considered a single gate in depth

Setting the Index Bit: The Index Bit here needs from 1-2 X-Gates to switch from 0 to 1 or two to switch from 0 to 1 then 1 to 0 depending on which index comes first for the select bits.

### Toffoli (CCX) Gate Resources Used

CCX Gate Decomposition is completely deterministic, so the number of resources used stays the same each time.

One CSWAP (controlled swap based on the index bit) is CNOT + Toffoli + CNOT, for a total of 8 2Qubit gates above.

```python
@partial(qp.compile, basis_set="CNOT")
@qp.qnode(qp.device("default.qubit"))
def toffoli_circuit():
    qp.Toffoli(wires=[0, 1, 2])
    return qp.state()

r = qp.specs(toffoli_circuit)()["resources"]

print("Depth:", r.depth)
print("1Q Gates:", r.gate_sizes.get(1, 0))
print("2Q Gates:", r.gate_sizes.get(2, 0))
print("Gate Types:", r.gate_types)
```

### Number of Gates for QROM_size = 4

With 2 bits used as index bits, we have three Toffoli (CCNOT) gates for each of the QROM slots (NOT INCLUDING THE CSWAP logic, so only 6 2Q gates instead of 8). One Toffoli has 13 1Q gates, 6 2Q gates, and 14 depth. x4 that will be 52 1Q gates and 24 2Q. 2 bits per value, so x2 on top of that is 104 1Q and 48 2Q. Then, to toggle the index bits, 4 X-gates are needed to flip 00 -> 01 -> 10 -> 11. However, we can eliminate CCNOT gates a lot if the target bits are set to 0, so that's why the mins are very low.

### MultiControlledX Gates

```python
def analyze_mcx(k):

    wires = list(range(k + 1))  # k controls + 1 target

    @partial(qp.compile, basis_set="CNOT")
    @qp.qnode(qp.device("default.qubit"))
    def circuit():
        qp.MultiControlledX(wires=wires)
        return qp.state()

    r = qp.specs(circuit)()["resources"]

    print(f"\n{k}-Controlled X")
    print("-" * 30)
    print("Controls :", k)
    print("Qubits   :", k + 1)
    print("Depth    :", r.depth)
    print("1Q Gates :", r.gate_sizes.get(1, 0))
    print("2Q Gates :", r.gate_sizes.get(2, 0))
    print("Gate Types:")
    for gate, count in r.gate_types.items():
        print(f"  {gate}: {count}")

    # qp.draw_mpl(circuit, style="pennylane")()
    # plt.show()

    return r
```

```python
analyze_mcx(k=10)
```
